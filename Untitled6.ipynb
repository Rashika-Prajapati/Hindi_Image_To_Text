```python
#Image preprocess
from PIL import Image, ImageFilter
import pytesseract

def preprocess_image(image_path):
    image = Image.open(image_path)
    # Convert to grayscale
    image = image.convert('L')
    image = image.filter(ImageFilter.MedianFilter(size=3))
    print(image)
    return image
```

```python
#Extract text from image
def extract_text(image_path):
    image = Image.open(image_path)
    text = pytesseract.image_to_string(image,lang='hin')
    print(text)
    return text
```

```python
#Normalize Text
import re
from indicnlp.normalize import indic_normalize

def normalize_text(text):
    text = text.lower()
    text = re.sub(r'[^\w\s]', '', text)
    normalizer = indic_normalize.IndicNormalizerFactory().get_normalizer('hi') # 'hi' for Hindi
    text = normalizer.normalize(text)
    return text
```

```python
#Tokenize text
from indicnlp.tokenize import indic_tokenize

def tokenize_text(text):
    tokens = indic_tokenize.trivial_tokenize(text)
    return tokens
```

```python
#language detection
from langdetect import detect

def detect_language(text):
    language = detect(text)
    return language
```

```python
image_path = '/content/Screenshot 2025-04-13 170010.png'
image = preprocess_image(image_path)
text = extract_text(image_path)
normalized_text = normalize_text(text)
tokens = tokenize_text(normalized_text)
language = detect_language(normalized_text)

print("Extracted Text:", text)
print("Normalized Text:", normalized_text)
print("Tokens:", tokens)
print("Detected Language:", language)
```

```python
output_file_path = '/content/output.txt'  # Or any desired path

# Open the file in write mode ('w')
with open(output_file_path, 'w', encoding='utf-8') as output_file:
    # Write the output to the file
    output_file.write("Extracted Text: " + text + "\n")
    output_file.write("Normalized Text: " + normalized_text + "\n")
    output_file.write("Tokens: " + str(tokens) + "\n")
    output_file.write("Detected Language: " + language + "\n")

print("Output saved to:", output_file_path)
```
